```python
# --- Merge all station metadata into one unified dictionary ---
stations_all = {}

# Add Salerno AWS
for sta in stations_info_salerno:
    stations_all[sta["ID"]] = (sta["Lat"], sta["Lon"], "SalernoAWS")

# Add Glacioclim AWS
for sta in stations_info_glacioclim:
    stations_all[sta["ID"]] = (sta["Lat"], sta["Lon"], "Glacioclim")

# Add NatGeo / Everest AWS
for sta in stations_info_everest:
    stations_all[sta["ID"]] = (sta["Lat"], sta["Lon"], "NatGeo")

# Check summary
print(f"Total stations merged: {len(stations_all)}")
print(stations_all.keys())
```

```python
lat,lon=LATS,LONS
# --------------------------
def nearest_ij(lat2d, lon2d, lat, lon):
    """Return nearest (j, i) indices in the model grid."""
    dist = np.hypot(lat2d - lat, lon2d - lon)
    j, i = np.unravel_index(np.nanargmin(dist), dist.shape)
    return j, i

# --------------------------
# Compute indices + collect summary
# --------------------------
nearest_indices = {}
rows = []

for name, (lat_sta, lon_sta, group) in stations_all.items():
    j, i = nearest_ij(lat, lon, lat_sta, lon_sta)
    nearest_indices[name] = (j, i)
    rows.append({
        "ID": name,
        "Group": group,
        "Station_Lat": lat_sta,
        "Station_Lon": lon_sta,
        "Grid_j": j,
        "Grid_i": i,
        "Model_Lat": float(lat[j, i]),
        "Model_Lon": float(lon[j, i]),
        "Model_Elev_m": float(dsSH.values[j, i])
    })

df_station_to_grid = pd.DataFrame(rows).sort_values(["Group", "ID"]).reset_index(drop=True)

print(df_station_to_grid.to_string(index=False))
```

```python
# --------------------------
# Salerno
jAWS0, iAWS0   = nearest_indices["AWS0"]
jAWS1, iAWS1   = nearest_indices["AWS1"]
jAWS2, iAWS2   = nearest_indices["AWS2"]
jAWS3, iAWS3   = nearest_indices["AWS3"]
jAWS4, iAWS4   = nearest_indices["AWS4"]
jAWS5, iAWS5   = nearest_indices["AWS5"]
jAWSSC, iAWSSC = nearest_indices["AWSSC"]
jCNG, iCNG     = nearest_indices["CNG_SNP"]
jPYR2, iPYR2   = nearest_indices["PYmon"]

# Glacioclim
jMN, iMN = nearest_indices["MN"]
jCH, iCH = nearest_indices["CH"]
jML, iML = nearest_indices["ML"]
jKH, iKH = nearest_indices["KH"]
jPH, iPH = nearest_indices["PH"]
jPY, iPY = nearest_indices["PY"]
jMS, iMS = nearest_indices["MS"]
jMC, iMC = nearest_indices["MC"]

# NatGeo (Everest)
jbc, ibc     = nearest_indices["bc"]
jc2, ic2     = nearest_indices["c2"]
jphort, iphort = nearest_indices["phort"]
jbish, ibish   = nearest_indices["bc"]
jsc, isc       = nearest_indices["sc"]

# Optional: save lookup table
df_station_to_grid.to_csv("/bettik/PROJECTS/pr-regional-climate/santolam/station_grid_lookup_MARnxny10.csv", index=False)
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Event dates
idate_event = "2013-10-10"
fdate_event = "2013-10-18"

# Assign a color per station
station_names = list(stations_rr_hourly_all.keys())
colors = plt.cm.tab10.colors  # 10 distinct colors from colormap
color_map = {name: colors[i % len(colors)] for i, name in enumerate(station_names)}
linestyle_sta=[":",":",":",
              "-","-","-","-","-","-",
              "-","-"]
lw_obs=2
lw_mod=3
fig, ax = plt.subplots(figsize=(18,10))

for name in station_names:
    obs = stations_rr_daily_all[name]
 #   model = ts_TF_model_all[name]
    color = color_map[name]

    if isinstance(obs, xr.DataArray):
        obs_event = obs.sel(time=slice(idate_event, fdate_event))
        x_obs = obs_event.time.values
        y_obs = obs_event.values
    elif isinstance(obs, pd.Series):
        obs_event = obs.loc[idate_event:fdate_event]
        x_obs = obs_event.index
        y_obs = obs_event.values
 
    else:
        raise TypeError(f"Unsupported type for station {name}: {type(obs)}")

    #obs_event = stations_rr_daily_all[name].sel(time=slice(idate_event, fdate_event))
    model_event = ts_TF_model_all[name].sel(time=slice(idate_event, fdate_event))

    color = color_map[name]
    # Plot observed 
   # ax.plot(obs_event.time, obs_event.values, 
    #       color=color, lw=2, alpha=0.8, linestyle='-',label=f"{name} (Observed)")
    ax.plot(model_event.time, y_obs, color=color, lw=3,linestyle=':', alpha=0.9, label=f"{name} (Obs)")
    # Plot model (line)
    ax.plot(model_event.time, model_event.values, 
            color=color, lw=3, alpha=0.8, linestyle='-',label=f"{name} (Model)")

    # Plot model (bars)
    #ax.plot(model_event.time, model_event, color='lightblue',  alpha=0.8, label=f"{name} (Model)")

# Labels and title
ax.set_xlabel("Date",fontsize=18)
ax.set_ylabel("Precipitation [mm/day]", fontsize=18)
#ax.set_title("Daily Total Precipitation – In situ (dotted) vs MAR (solid)", fontsize=18)#, fontweight='bold')
ax.grid(True, linestyle='--', alpha=0.4)

ax.tick_params(axis='both', labelsize=16)




# --- Custom legend ---
legend_elements = []

# Add one colored square per station
for name in station_names:
    legend_elements.append(Line2D([0], [0], marker='s', color='w',
                                  markerfacecolor=color_map[name], markersize=12,
                                  label=name))

# Add dotted Obs and solid MAR line examples
legend_elements.append(Line2D([0], [0], color='grey', lw=lw_obs, linestyle=':', label='Station'))
legend_elements.append(Line2D([0], [0], color='black', lw=lw_mod, linestyle='-', label='MAR'))

ax.legend(handles=legend_elements, fontsize=16, frameon=True, loc='upper left', ncol=2)

plt.tight_layout()
plotname=f'FIGX_pre_hourly_event_{idate_event}_all_stations'
plotsDir = "/home/santolam/RECAP/figures/paper_on_validation_RECAP/"
#plt.savefig(f"{plotsDir}{plotname}.png", bbox_inches="tight", dpi=300)
#print(f"Saved: {plotsDir}{plotname}.png")


plt.show()
```
