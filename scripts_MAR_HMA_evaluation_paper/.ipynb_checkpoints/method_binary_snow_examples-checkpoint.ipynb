```python
##esacciv20
sourceData='/bettik/PROJECTS/pr-regional-climate/santolam/esacciv20/HMA/'
path_out=sourceData+'snow_bin'

for year in np.arange(1982,2018+1,1):
    ds=xr.open_mfdataset(sourceData+'/daily/'+
                    f'{year}*-ESACCI-L3C_SNOW-SCFG-AVHRR_MERGED-fv2.0_interp_HMA.nc')['scfg_interp']
    ds_SNC=ds > 10
    ds_SNC= ds_SNC.astype(np.uint8)
    binary_SNC = xr.DataArray(
        data=ds_SNC.astype('uint8'),  # convert to smaller binary dtype
        coords={
            'time': ds['time'],
            'lat': ds['lat'],
            'lon': ds['lon']
        },
        dims=('time', 'lat', 'lon'),
        attrs={
            'long_name': 'Binary Snow Cover Fraction',
            'units': '1',  # or 'binary flag'
            'description': '1 = Snow-covered, 0 = No snow',
            'source': 'Derived from original Snow Cover Fraction data',

        }
    )
    binary_SNC.to_netcdf(sourceData+'/snow_bin/'+'snc_bin_'+str(year)+'-ESACCI-L3C_SNOW-SCFG-AVHRR_MERGED-fv2.0_interp_HMA.nc')
```

```python

##MAR GRp
##
path_out_main = '/bettik/PROJECTS/pr-regional-climate/santolam/MARout_post/GRp/spin2/work/snow_bin/'
os.makedirs(path_out_main, exist_ok=True)
path_out=path_out_main

sourceData='/bettik/PROJECTS/pr-regional-climate/santolam/MARout_post/GRp/spin2/work/'
#fileName=str(year)+'*-ESACCI-L3C_SNOW-SCFG-AVHRR_MERGED-fv2.0_interp_HMA'
#year=1994
for year in np.arange(1950,2024+1,1):
    ds=xr.open_dataset(sourceData+'/daily/'+
                    f'SHSN3_daymean_MARv3.14_ER5_spin2_GRp_{year}.nc')['SHSN3'].sel(SECTOR=2)
    ds_SNC=ds > 0.1
    ds_SNC= ds_SNC.astype(np.uint8)
    binary_SNC = xr.DataArray(
        data=ds_SNC.astype('uint8'),  # convert to smaller binary dtype
            coords={
        'TIME': ds['TIME'],
        'Y': ds['Y'],
        'X': ds['X']
    },
    dims=('TIME', 'Y', 'X'),
    attrs={
        'long_name': 'Binary Snow Cover Fraction',
        'units': '1',  # or 'binary flag'
        'description': '1 = Snow-covered, 0 = No snow',
        'source': 'Derived from original SHSN3 > 0.1m',

    }
    )
    binary_SNC.to_netcdf(path_out+f'SHSN3_bin_daymean_MARv3.14_ER5_spin2_GRp_{year}.nc')
```
